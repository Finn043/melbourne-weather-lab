# Melbourne Airport weather: exploratory analysis
Load the shared NOAA preprocessing pipeline, inspect coverage and missing values, then visualize daily and seasonal conditions. Run from the project root after installing `requirements.txt`.


```python
import matplotlib.pyplot as plt
from src.preprocess import prepare_weather_data
weather = prepare_weather_data()
weather.head()
```

```python
print(f'Records: {len(weather):,}')
print(f'Coverage: {weather.index.min().date()} — {weather.index.max().date()}')
weather[['tmin', 'tmax', 'prcp']].describe().round(2)
```

```python
fig, axes = plt.subplots(2, 1, figsize=(14, 8), sharex=True)
weather[['tmin', 'tmax']].resample('M').mean().plot(ax=axes[0], color=['#59a994', '#e5a842'])
axes[0].set_ylabel('Monthly mean temperature (°C)')
weather['prcp'].resample('M').sum().plot(ax=axes[1], color='#4b9f88')
axes[1].set_ylabel('Monthly precipitation (mm)')
plt.tight_layout()
```

```python
seasonal = weather.groupby(weather.index.month)[['tmin', 'tmax', 'prcp']].mean()
seasonal.index.name = 'Month'
seasonal.round(2)
```
